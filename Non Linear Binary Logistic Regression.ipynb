```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv('ushape.csv')
```

```python
df.head()
```

```python
x=df.iloc[:,:2]
y=df.iloc[:,-1]
```

```python
x
```

```python
y.dtype
```

```python
import matplotlib.pyplot as plt
```

```python
plt.scatter(x.iloc[:,0],x.iloc[:,-1],c=y)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
lr=LogisticRegression()
```

```python
lr.fit(x,y)
```

```python
from mlxtend.plotting import plot_decision_regions
```

```python
plot_decision_regions(x.values,y.values.astype('int'),lr)
```

```python
from sklearn.model_selection import cross_val_score
np.mean(cross_val_score(lr,x,y,scoring='accuracy',cv=10))
```

```python
from sklearn.preprocessing import PolynomialFeatures
```

```python
poly=PolynomialFeatures(degree=3)
```

```python
x_poly=poly.fit_transform(x)
```

```python
lr1=LogisticRegression()
```

```python
lr1.fit(x_poly,y)
```

```python
from sklearn.model_selection import cross_val_score
np.mean(cross_val_score(lr1,x_poly,y,scoring='accuracy',cv=10))
```

```python
from sklearn.pipeline import make_pipeline
```

```python
def poly_log_reg(dg):
    model=make_pipeline(
        PolynomialFeatures(degree=dg),
        LogisticRegression(max_iter=1000)
    )
    model.fit(x.values,y.values.astype('int'))
    plot_decision_regions(x.values, y.values.astype('int'), model)
    plt.xlabel('Feature 1')
    plt.ylabel('Feature 2')
    plt.show()
    return np.mean(cross_val_score(model,x,y,scoring='accuracy',cv=10))
```

```python
poly_log_reg(10)
```
